```python
#K-MEANS

from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.metrics import davies_bouldin_score, silhouette_score
import pandas as pd
import numpy as np
# Load the Iris dataset
# Load Iris dataset (all feature columns)
data = pd.read_csv('iris.txt', sep=',', header=None)
data = data[[0,1,2,3]]
#data=data.dropna(subset=[6])
X = data.values

# Perform KMeans clustering with k=3
kmeans = KMeans(n_clusters=3,max_iter=200,tol=0)
labels = kmeans.fit_predict(X)

# Display cluster centers and labels
#print("\nCluster Centers:\n", kmeans.cluster_centers_)
print(np.array2string(kmeans.cluster_centers_,  separator=','))
#print("Labels:\n", labels)
print(np.array2string(labels,  separator=','))
print("Inertia:", kmeans.inertia_)

# Calculate Davies-Bouldin Score
db_score = davies_bouldin_score(X, labels)

# Calculate Silhouette Score
sil_score = silhouette_score(X, labels)

# Print the results
print(f"Davies-Bouldin Score: {db_score:.13f}")
print(f"Silhouette Score: {sil_score:.13f}")

print("labels = [", end="")
for i in range(len(labels)):
    end_char = ", " if i < len(labels) - 1 else "]"
    #print(labels[i], end=end_char)

```
