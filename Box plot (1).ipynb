```python
#Box plot 
import matplotlib.pyplot as plt
import pandas as pd
#from sklearn.datasets import load_iris

X = pd.read_csv('iris.txt',sep=',',header=None)
X = X[[0,1,2,3,4]]
feature_names = [
    'f1',
    'f2',
    'f3',
    'f4',
    'Class'
]

# Assign feature names to DataFrame columns
X.columns = feature_names

# Create box plots
plt.figure(figsize=(10, 6))
X.boxplot()

# Labels and title
plt.xlabel("Features")
plt.ylabel("Feature Ranges")
plt.title("IRIS Dataset")

# Rotate x-axis labels for better readability
plt.xticks(rotation=20)

# Show plot
plt.tight_layout()
plt.savefig("C:\\Users\\hp\Desktop\\Iris.jpg", format='jpg', dpi=400)
plt.show()
```
